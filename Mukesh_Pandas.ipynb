##### Q1. NumPy broadcasting

````python
import numpy as np

a = np.array([[10], [20], [30]]) 
b = np.array([[1, 2, 3]])         
result = a + b
print("Result:\n", result)
print("Shape:", result.shape)

b2 = np.array([1, 2])
try:
    result2 = a + b2
    print("\nWith b = np.array([1, 2]):")
    print(result2)
    print("Shape:", result2.shape)
except ValueError as e:
    print("\nValueError with b = np.array([1, 2]):", e)
````

**Predicted Output:**
```
 [[11 12 13]
 [21 22 23]
 [31 32 33]]
Shape: (3, 3)
```

**Explanation:**

1. `a` has shape (3, 1) and `b` has shape (1, 3). NumPy compares shapes right-to-left and stretches any dimension of size 1 to match the other, yielding a (3, 3) output — an outer-sum of all row–column pairs.
2. This differs from adding two 1-D arrays of shape (3,), where elements are matched positionally (index-to-index), giving only 3 values instead of 9.
3. With `b = np.array([1, 2])` (shape (2,)), NumPy aligns it as (1, 2) against `a`'s (3, 1): dimension 0 stretches `b` → 3, dimension 1 stretches `a` → 2, producing shape (3, 2) with no error.
4. A `ValueError` would only occur if neither dimension is 1 (e.g., `a` shape (3,1) + `b` shape (2,2)), since 3 and 2 are incompatible and cannot be broadcast.


##### Q2. CSV parsing, missing values, and dates

````python
import pandas as pd
from io import StringIO

csv_text = '''ID,Name,Date,Marks
101,"Asha, R",2026-09-01,80
102,Ravi,01/09/2026,
103,Neha,2026-09-03,NA
'''

df = pd.read_csv(
    StringIO(csv_text),
    na_values=["NA", ""],         
    keep_default_na=True
)
print("── Raw load ──")
display(df)
print()
print("dtypes after raw load:")
display(df.dtypes)

df["Date"] = pd.to_datetime(df["Date"], dayfirst=False, format="mixed")

print("\n── Missing Marks (investigate before dropping) ──")
display(df[df["Marks"].isna()])

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print("\n── Cleaned DataFrame ──")
display(df)
print()
print("dtypes after cleaning:")
display(df.dtypes)
````

**Predicted Output:**
```
    ID     Name        Date  Marks
0  101  Asha, R  2026-09-01   80.0
1  102     Ravi  2026-01-09    NaN
2  103     Neha  2026-09-03    NaN

dtypes after cleaning:
ID                int64
Name                str
Date     datetime64[us]
Marks           float64
```

**Explanation:**

1. `pd.read_csv()` follows the RFC 4180 standard and treats text inside double-quotes as a single field, so the comma in `"Asha, R"` is not split — it is read correctly as one value in the Name column.
2. The Date column contains two formats (YYYY-MM-DD and DD/MM/YYYY); without `pd.to_datetime(..., format="mixed")` the column stays as plain string (object dtype), making date arithmetic and filtering impossible.
3. Both a blank entry and the string "NA" are converted to NaN using `na_values=["NA", ""]`; however, "NA" must be investigated before treating it as missing — it may mean "Not Assessed" or "Not Applicable" rather than a genuinely absent value, and dropping it blindly could remove a valid record.
4. After parsing, missing Marks are filled with the column mean as a safe placeholder; in practice the correct action (fill, drop, or flag) should be decided only after understanding why each value is missing.


##### Q3. Salary outlier and robust summary

````python
import pandas as pd

df = pd.DataFrame({
    "Employee": ["A", "B", "C", "D"],
    "Salary":   [25000, 30000, 35000, 10000000]
})

mean_salary   = df["Salary"].mean()
median_salary = df["Salary"].median()

print(f"Mean   Salary : {mean_salary:,.2f}")
print(f"Median Salary : {median_salary:,.2f}")

highest_paid = df[df["Salary"] == df["Salary"].max()]
print("\n── Highest-paid employee record ──")
display(highest_paid)

mean_without_outlier = df[df["Salary"] < 10000000]["Salary"].mean()
print(f"\nMean without outlier : {mean_without_outlier:,.2f}")
print(f"Mean with    outlier : {mean_salary:,.2f}")
print(f"Difference           : {mean_salary - mean_without_outlier:,.2f}")

print(f"\nMedian better represents a typical salary: {median_salary:,.2f}")
````

**Predicted Output:**
```
Mean   Salary : 2,522,500.00
Median Salary :    32,500.00

Highest-paid employee record:
  Employee    Salary
3        D  10000000

Mean without outlier : 30,000.00
Mean with    outlier : 2,522,500.00
Difference           : 2,492,500.00

Median better represents a typical salary: 32,500.00
```

**Explanation:**

1. The mean is (25000 + 30000 + 35000 + 10000000) / 4 = 2,522,500 and the median is the average of the two middle values (30000 + 35000) / 2 = 32,500.
2. The extreme salary of 10,000,000 pulls the mean up by over 2.4 million, making it far higher than three of the four actual salaries and an unreliable measure of central tendency for this dataset.
3. The median (32,500) is the better representative of a typical employee's salary because it is unaffected by extreme values and sits within the realistic range of the other three salaries (25,000–35,000).
4. Employee D's record must be investigated before removal — the value could be a legitimate executive salary, a data-entry error (extra zeros), or a unit mismatch (e.g., rupees vs dollars); removing it without verification risks deleting a real and important record.

##### Q1. Clean student assessment data 

````python
import pandas as pd
import numpy as np

df = pd.read_csv("student_assessment.csv", dtype=str)
print("-- head() --")
display(df.head())
print("\n-- info() --")
df.info()
print("\n-- isnull().sum() --")
print(df.isnull().sum())
rows_loaded = len(df)

dept_map = {
    "it":      "IT",
    "i.t.":    "IT",
    "finance": "Finance",
    "hr":      "HR",
}
df["Department"] = df["Department"].str.strip().str.lower().map(dept_map)
print("\n-- Department value counts after standardization --")
print(df["Department"].value_counts())

before_dedup = len(df)
df = df.drop_duplicates()
duplicates_removed = before_dedup - len(df)
print(f"\nDuplicates removed : {duplicates_removed}")

df["Absent_Flag"] = df["Marks"].str.strip().str.lower() == "absent"

df["Marks"] = (
    df["Marks"]
    .str.extract(r"(\d+\.?\d*)", expand=False)
)
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

print("\n-- Marks after parsing --")
display(df[["Student_ID", "Student_Name", "Marks", "Absent_Flag"]])

absent_students = df[df["Absent_Flag"]]
unentered_marks = df[~df["Absent_Flag"] & df["Marks"].isna()]

print(f"\nAbsent students ({len(absent_students)}):")
display(absent_students[["Student_ID", "Student_Name"]])

print(f"\nUnentered marks ({len(unentered_marks)}):")
display(unentered_marks[["Student_ID", "Student_Name"]])

invalid_marks = df[df["Marks"].notna() & ((df["Marks"] < 0) | (df["Marks"] > 100))]
print(f"\nInvalid marks outside 0-100 ({len(invalid_marks)}):")
display(invalid_marks[["Student_ID", "Student_Name", "Marks"]])

print("\n== Cleaned DataFrame ==")
display(df[["Student_ID", "Student_Name", "Department", "Marks", "Absent_Flag"]])

print("\n== Summary Report ==")
print(f"  Records loaded              : {rows_loaded}")
print(f"  Duplicates removed          : {duplicates_removed}")
print(f"  Records after dedup         : {len(df)}")
print(f"  Absent students             : {len(absent_students)}")
print(f"  Unentered marks (NaN)       : {len(unentered_marks)}")
print(f"  Invalid marks (outside 0-100): {len(invalid_marks)}")
````

**Cleaning Decisions, Results, and Justification:**

1. **Load & Inspect** — The dataset is loaded with `dtype=str` so every value is read as plain text, preventing Pandas from silently converting `"88 marks"` or `"Absent"` to `NaN`. `head()`, `info()`, and `isnull().sum()` are used to inspect structure and missing values before cleaning.

2. **Standardize Departments** — A lowercase-lookup dictionary maps `it`, `i.t.` → `IT` and `finance` → `Finance` using `.str.lower().map(dept_map)`, making the logic explicit and consistent across all casing variants.

3. **Remove Duplicates** — `drop_duplicates()` removes rows where every column is identical. The record `(S102, Ravi, it, Python, 78)` appeared twice and was reduced to one; records sharing a Student_ID but with different marks are retained.

4. **Parse Marks** — `"88 marks"` is cleaned using regex to extract the numeric part; `"Absent"` is flagged in an `Absent_Flag` column before conversion; `"NA"` and blanks become `NaN` via `pd.to_numeric(..., errors="coerce")`.

5. **Absence vs Unentered** — S105 (Meera) is marked `Absent` (did not sit the assessment); S104 (Karan), S109 (Sneha), and S112 (Aman) have blank entries (marks not yet recorded). **Risk:** Dropping all `NaN` rows without investigation silently removes absent students, shrinking the cohort and making pass rates appear artificially higher.

6. **Out-of-Range Marks** — S114 (Kabir) has a mark of `105`, which exceeds the valid range of 0–100. This must be verified against the source record before correction or removal — it could be a data-entry error (e.g., `95` mistyped as `105`).
